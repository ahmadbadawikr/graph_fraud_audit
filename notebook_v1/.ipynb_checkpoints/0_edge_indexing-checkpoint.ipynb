# 0. Edge Indexing

This notebook reads edge lists from Parquet files, maps the source and destination node IDs to the contiguous integer IDs (generated in Step 0b), and stores the result in an LMDB database. 
This indexed format allows for extremely fast random access during subsequent graph construction steps.

```python
# Configuration
import os

# EXTERNAL DRIVE CONFIGURATION
ROOT_DIR = "/Volumes/Backup Plus/Zaman/graph"
DATA_DIR = os.path.join(ROOT_DIR, "data")
MAPPING_DIR = os.path.join(ROOT_DIR, "lmdb_node_mapping")
OUTPUT_INDEX_DIR = os.path.join(ROOT_DIR, "lmdb_edge_indexing")

# Map Map Size (Bytes) for Output LMDBs
DEFAULT_MAP_SIZE = 1024 * 1024 * 1024 * 5 # 5GB Default

# Edge Configuration
# format: "EdgeName": { "src": "NodeMapDesc", "dst": "NodeMapDesc", "folder": "ParquetFolder" }
# NodeMapDesc can be a string (filename) or a list of strings (if nodes span multiple maps)
EDGES_CONFIG = {
    "edge_rek_credit": {
        "src": "transaksi.lmdb",
        "dst": ["simpanan.lmdb", "pinjaman.lmdb"], # Destination can be either simpanan or pinjaman
        "folder": "edge_rek_credit",
        "src_col": "src",
        "dst_col": "dst"
    },
    "edge_rek_debit": {
        "src": ["simpanan.lmdb", "pinjaman.lmdb"],
        "dst": "transaksi.lmdb",
        "folder": "edge_rek_debit",
        "src_col": "src",
        "dst_col": "dst"
    },
    # Note: If you ran Step 0a (Split Edges), you might want to index the split edges instead:
    # Uncomment below if you want to use the split edges
    # "edge_pinj_credit": {
    #     "src": "transaksi.lmdb",
    #     "dst": "pinjaman.lmdb",
    #     "folder": "edge_pinj_credit",
    #     "src_col": "src",
    #     "dst_col": "dst"
    # },
    # ... and so on
}

os.makedirs(OUTPUT_INDEX_DIR, exist_ok=True)
```

```python
# Imports
import lmdb
import pyarrow as pa
import pyarrow.dataset as ds
from tqdm.notebook import tqdm
import shutil
```

```python
# Helpers 

def normalize_key(v):
    """Normalizes key to match mapping format (strips .0, etc.)"""
    if v is None:
        return None
    s = str(v).strip()
    if s.endswith(".0"):
        s = s[:-2]
    return s

def open_lmdb_readonly(path):
    return lmdb.open(path, readonly=True, lock=False)

def lookup_single(env, key):
    if key is None: return None
    k = normalize_key(key).encode()
    with env.begin() as txn:
        v = txn.get(k)
        # Return decoded string ID
        return v.decode() if v else None

def lookup_multi(envs, key):
    if key is None: return None
    k = normalize_key(key).encode()
    for env in envs:
        with env.begin() as txn:
            v = txn.get(k)
            if v:
                return v.decode()
    return None

def patch_schema(schema):
    fields = []
    for field in schema:
        if pa.types.is_decimal(field.type):
            fields.append(pa.field(field.name, pa.string()))
        else:
            fields.append(field)
    return pa.schema(fields)
```

```python
# Indexing Function
def index_edge(edge_name, config):
    print(f"\n=== Indexing {edge_name} ===")
    
    # 1. Prepare Source Mappings
    src_map_conf = config["src"]
    if isinstance(src_map_conf, list):
        src_envs = [open_lmdb_readonly(os.path.join(MAPPING_DIR, f)) for f in src_map_conf]
        lookup_src = lambda k: lookup_multi(src_envs, k)
    else:
        env_src = open_lmdb_readonly(os.path.join(MAPPING_DIR, src_map_conf))
        lookup_src = lambda k: lookup_single(env_src, k)
        
    # 2. Prepare Dest Mappings
    dst_map_conf = config["dst"]
    if isinstance(dst_map_conf, list):
        dst_envs = [open_lmdb_readonly(os.path.join(MAPPING_DIR, f)) for f in dst_map_conf]
        lookup_dst = lambda k: lookup_multi(dst_envs, k)
    else:
        env_dst = open_lmdb_readonly(os.path.join(MAPPING_DIR, dst_map_conf))
        lookup_dst = lambda k: lookup_single(env_dst, k)

    # 3. Prepare Input Dataset
    folder_path = os.path.join(DATA_DIR, config["folder"])
    if not os.path.exists(folder_path):
        print(f"Error: Folder {folder_path} does not exist. Skipping.")
        return

    try:
        base = ds.dataset(folder_path, format="parquet", partitioning="hive")
        schema = patch_schema(base.schema)
        dataset = ds.dataset(folder_path, format="parquet", partitioning="hive", schema=schema)
    except Exception as e:
        print(f"Error opening dataset: {e}")
        return

    # 4. Prepare Output LMDB
    out_path = os.path.join(OUTPUT_INDEX_DIR, f"{edge_name}.lmdb")
    env_out = lmdb.open(out_path, map_size=DEFAULT_MAP_SIZE)

    # 5. Process
    scanner = dataset.scanner(columns=[config["src_col"], config["dst_col"]])
    
    total_edges = 0
    matched = 0
    missed = 0
    
    # Transaction for writing
    txn = env_out.begin(write=True)
    
    pbar = tqdm(desc=f"{edge_name}")
    
    for batch in scanner.to_batches():
        batch_dict = batch.to_pydict()
        src_list = batch_dict[config["src_col"]]
        dst_list = batch_dict[config["dst_col"]]
        
        pbar.update(len(src_list))
        
        for s_raw, d_raw in zip(src_list, dst_list):
            sid = lookup_src(s_raw)
            did = lookup_dst(d_raw)
            
            if sid is not None and did is not None:
                matched += 1
                # Write Key: edge_index, Value: "src_id,dst_id"
                # Note: The original code used a simple counter 'edge_id' as the key.
                # This is sufficient for sequential reading.
                txn.put(str(total_edges).encode(), f"{sid},{did}".encode())
                total_edges += 1
            else:
                missed += 1
        
        # Commit periodically
        if matched % 100_000 == 0:
            txn.commit()
            txn = env_out.begin(write=True)
            
    txn.commit()
    env_out.close()
    pbar.close()
    
    print(f"Finished {edge_name}.")
    print(f"Total Matched: {matched:,}")
    print(f"Total Missed : {missed:,}")
    print(f"Saved to {out_path}")
```

```python
# Run Indexing for all configured edges
for edge, cfg in EDGES_CONFIG.items():
    index_edge(edge, cfg)
```
